La práctica usa el título como texto y una etiqueta didáctica `alta_calificacion = average_rating >= 4.2`. El texto no contiene la calificación ni ninguna variable que revele directamente la respuesta.

```python
import sys
from pathlib import Path

ROOT = Path.cwd().resolve()
while ROOT != ROOT.parent and not (ROOT / "src").exists():
    ROOT = ROOT.parent
sys.path.insert(0, str(ROOT / "src"))

from hf_loader import HFDatasetLoader
from transformer_goodreads import prepare_text_data, fit_transformer

raw = HFDatasetLoader("pfaha/goodreads-books", split="train").load()
texts = prepare_text_data(raw)
result = fit_transformer(texts["texto"], texts["alta_calificacion"])
print(result["metrics"])
```
