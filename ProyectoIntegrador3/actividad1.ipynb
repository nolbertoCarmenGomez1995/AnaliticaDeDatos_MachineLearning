```python
import sys
from pathlib import Path

# Agregar la carpeta src al sys.path
ROOT = Path.cwd().resolve()
sys.path.insert(0, str(ROOT / "src"))

from hf_loader import HFDatasetLoader
from reportes import Reportes
```

```python
# Instanciar la clase de carga con tu dataset de libros de fantasía
loader = HFDatasetLoader(
    dataset_name="AlekseyKorshuk/fantasy-books",
    split="train",
)

# Cargar el DataFrame original
df_original = loader.load()

# Mostrar la vista previa
df_original.head()
```

```python
# Verificamos las columnas disponibles en el dataset
print("Columnas disponibles:", df_original.columns.tolist())

# Transformación mínima: Seleccionar columnas principales y renombrarlas al español
columnas_deseadas = [col for col in ["title", "author", "rating", "description"] if col in df_original.columns]

if columnas_deseadas:
    df_transformado = df_original[columnas_deseadas].copy()
else:
    df_transformado = df_original.iloc[:, :5].copy()

# Renombrar columnas seleccionadas
df_transformado = df_transformado.rename(columns={
    "title": "titulo",
    "author": "autor",
    "rating": "calificacion",
    "description": "descripcion"
})

print(f"Dimensiones tras la transformación: {df_transformado.shape}")
df_transformado.head()
```

### Explicación de la Transformación Realizada
Se realizó una **transformación mínima** consistente en:
1. **Selección de variables:** Se extrajeron únicamente las columnas clave del catálogo (`title`, `author`, `rating`, `description`).
2. **Renombrado de columnas:** Se traducen sus identificadores al español (`titulo`, `autor`, `calificacion`, `descripcion`) para facilitar la interpretación del análisis descriptivo.
